```python
import pandas as pd
import numpy as np

# Lê o arquivo CSV e guarda como DataFrame (tabela)
df_analise =pd.read_csv("retail_store_sales.csv")
display(df_analise)
```

```python
# Mostra o tipo de cada coluna e a quantidade de valores não nulos
display(df_analise.info())
```

```python
# Converte a coluna de data de texto para datetime
# errors='coerce' transforma datas inválidas em NaT (data vazia), sem dar erro
df_analise["Transaction Date"] = pd.to_datetime(df_analise["Transaction Date"], errors='coerce')

# confere se a conversão funcionou
display(df_analise.info())
```

```python
# Lista os valores únicos da coluna, para saber o que ela contém antes de converter
display(df_analise["Discount Applied"].unique())

# Converte de object para boolean (True/False), aceitando valores vazios
df_analise["Discount Applied"] = df_analise["Discount Applied"].astype("boolean")

# Confere o resultado
display(df_analise.info())
```

```python
# Converte Quantity de float para inteiro (Int64 com "I" maiúsculo aceita vazios)
df_analise["Quantity"] = df_analise["Quantity"].astype("Int64")

# Confere o resultado
display(df_analise.info())
```

```python
# Remove as linhas em que a coluna Item está vazia
df_analise = df_analise.dropna(subset=["Item"])

# Confere o resultado
display(df_analise)

# Lista os valores únicos da coluna, para saber o que ela contém antes de converter
display(df_analise["Item"].unique())

# Conta quantas vezes cada item aparece (dropna=False inclui os vazios na contagem)
df_analise["Item"].value_counts(dropna=False)
```

```python
# Confere o resultado
display(df_analise.info())

# Confere de novo se ainda existe algum item vazio
df_analise["Item"].value_counts(dropna=False)

```

```python
display(df_analise)

# Conta os valores de preço, incluindo vazios, para achar preços fora do padrão.
df_analise["Price Per Unit"].value_counts(dropna=False)
```

```python

# Lista das colunas numéricas que serão analisadas juntas
colunas = ["Price Per Unit", "Quantity", "Total Spent"]
 
# Mostra o tipo de dado de cada coluna da lista
display(df_analise[colunas].dtypes)
 
# Conta quantos valores vazios existem em cada coluna da lista
display(df_analise[colunas].isna().sum())
 
# Mostra as 15 primeiras linhas que têm algum valor vazio nessas colunas
display(
    df_analise.loc[
        df_analise[colunas].isna().any(axis=1),
        colunas
    ].head(15)
)
```

```python
# Conta True, False e vazios na coluna Discount Applied
display(df_analise["Discount Applied"].value_counts(dropna=False))

# Mostra o total de valores ausentes na coluna
display( 
    "Valores ausentes:",
    df_analise["Discount Applied"].isna().sum()
)
```

```python
# Preenche os vazios com False (assumindo que não houve desconto)
df_analise["Discount Applied"] = (
    df_analise["Discount Applied"].fillna(False)
)

# Confere se não sobrou nenhum valor vazio e como ficou a contagem
display(df_analise["Discount Applied"].value_counts(dropna=False))
```

```python
# Verificação final do DataFrame após todas as alterações
display(df_analise.info())

# Salva o DataFrame limpo em CSV (index=False evita criar uma coluna extra de índice)
df_analise.to_csv("retail_store_sales_limpo.csv", index=False)
```
